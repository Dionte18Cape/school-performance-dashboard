```python
# Average Student Grade by School
df = pd.read_csv('classroom_students_cleaned.csv')
students = pd.read_csv('students_cleaned.csv')
people = pd.read_csv('people_cleaned.csv')
schools = pd.read_csv('schools_cleaned.csv')

merged = df.merge(students, on='STUDENT_ID') \
           .merge(people, on='PERSON_ID') \
           .merge(schools, on='SCHOOL_ID')

avg_grade_school = merged.groupby('SCHOOL_NAME')['GRADE'].mean().round(2)
print(avg_grade_school)
```

```python
# Top 3 Performing Grade Levels
top_levels = merged.groupby('GRADE_LEVEL')['GRADE'].mean().sort_values(ascending=False).head(5)
print(top_levels)
```

```python
# Student with the Highest Grade
top_student = merged.loc[merged['GRADE'].idxmax(), ['FIRST_NAME', 'LAST_NAME', 'GRADE', 'SCHOOL_NAME']]
print(top_student)
```

```python
# Student-to-Teacher Ratio
student_count = students['STUDENT_ID'].nunique()
teacher_count = pd.read_csv('teachers_cleaned.csv')['TEACHER_ID'].nunique()
ratio = round(student_count / teacher_count, 2)
print(f"Student-to-Teacher Ratio: {ratio}")
```

```python

import pandas as pd

# Load the cleaned CSV files
principals = pd.read_csv('principals_cleaned.csv')
schools = pd.read_csv('schools_cleaned.csv')
people = pd.read_csv('people_cleaned.csv')

# Inspect column names to ensure they match
print("Principals Columns:", principals.columns)
print("Schools Columns:", schools.columns)
print("People Columns:", people.columns)

# Perform merges with left joins to preserve principal records even if school or person is missing
merged = principals.merge(schools, on='PRINCIPAL_ID', how='left') \
                   .merge(people, on='PERSON_ID', how='left')

# Inspect nulls to see where merges failed
print(merged[['PRINCIPAL_ID', 'PERSON_ID', 'SCHOOL_NAME', 'SALARY']].isnull().sum())

# Preview the top rows
print(merged.head())

# Extract and sort the final result
top_principals = merged[['FIRST_NAME', 'LAST_NAME', 'SCHOOL_NAME', 'SALARY']] \
                    .sort_values(by='SALARY', ascending=False)

print(top_principals)
```
